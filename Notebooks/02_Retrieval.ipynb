```python
# Load environment variable
from dotenv import load_dotenv
load_dotenv()
```

```python
###create a embedding by using langchain hugging face.

from langchain_huggingface import HuggingFaceEmbeddings

embedding_model = HuggingFaceEmbeddings(
        model = "BAAI/bge-large-en-v1.5",
        model_kwargs = {"device":"cpu"},
        encode_kwargs = {"normalize_embeddings":True}
)
```

```python
from langchain_chroma import Chroma

vectorstore = Chroma(
    persist_directory="../vectorstore/kubernetes_rag",
    collection_name="kubernetes_rag",
    embedding_function=embedding_model
)
```

```python
print("Total Number Of Vectors:",vectorstore._collection.count())
```

```python
user_query = "What is a Kubernetes Deployment?"
```

```python
## similarity search

docs = vectorstore.similarity_search(user_query,k=5)
docs
```

```python
for i,doc in enumerate(docs,start = 1):

    print(f"------- Document: {i}-------")
    print("Page Content:",doc.page_content[:500])
    print("Source:",doc.metadata.get("source"))
    print("Page:",doc.metadata.get("page"))
```

```python
### Similarity Score
score_docs = vectorstore.similarity_search_with_score(user_query,k=5)
score_docs
```

```python
for i,doc in enumerate(score_docs,start = 1):

    print(f"------- Document: {i}-------")
    print("Page Content:",doc[0].page_content[:500])
    print("Source:",doc[0].metadata.get("source"))
    print("Page:",doc[0].metadata.get("page"))
    print("Score:",doc[1])
```

```python
queries = [
    "What is a Kubernetes Deployment?",
    "What is a Kubernetes Pod?",
    "What is a Kubernetes Service?",
    "What is a ReplicaSet?",
    "What is a ConfigMap?"
]

for query in queries:
    results = vectorstore.similarity_search_with_score(query,k=3)
    print("Query:",query)
    for i,(doc,score) in enumerate(results,start=1):

        print("------ Document:{i} -------")
        print("Page Content:",doc.page_content[:500])
        print("Source:",doc.metadata.get("source"))
        print("Page:",doc.metadata.get("page"))
        print("Score:",score)
```
